## <font color=cornflowerblue>Model 4: Graphing</font>
### <font color=green>A jupyter notebook can be used to visually present the relationship betweeen two quantities.

Running the code cell below creates a plot showing the relationship between time and the mass of dry chick peas that are soaking up water (This data was generated when carrying out the POGIL-PCL experiment: "Can we make stew with these beans?").

```python
# Assign arrays corresponding to x and different y's

Time = np.array([0,2.5,5,7.5,10,20,30,60])

Water = np.array([10.293,11.044,11.746,12.093,12.505,13.133,13.689,14.672])

NaCl3M = np.array([9.611,10.174,10.387,10.672,10.903,11.433,11.745,13.415])

# Plot data
# Marker options here: https://matplotlib.org/stable/api/markers_api.html
# Color options here: https://matplotlib.org/stable/gallery/color/named_colors.html
# Examples of plot types here: https://matplotlib.org/stable/gallery/index.html

plt.plot(Time,Water,marker="o",linestyle='None')

plt.plot(Time,NaCl3M,marker='x',color='r',linestyle='None')

plt.xlabel("Time (min)")

plt.ylabel("Mass (g)");

plt.legend(['Water',  '3M NaCl'], loc='lower right')
```

## <font color=cornflowerblue>Model 5: Functions</font>
### <font color=green> A jupyter notebook can execute a number of common mathematical functions
We have already used the NumPy array function and we will be using more numpy functions in the following sections.  NumPy also provides many common mathematical functions that can be applied to arrays of numbers.  This link contains more information about math functions that are available in numpy:
https://numpy.org/doc/stable/reference/routines.math.html.

```python
# Assign a variable a value

C0 = 1.0           # Initial concentration (M)
k = 1.0E-3         # 1st order rate constant (1/s)
time =20           # Time (s)

# Assign a variable using functions and operations of assigned variables

Conc = C0*np.exp(-k*time)

lnConc = np.log(C0)-k*time    # np.log() is the numpy natural log

# Display variables

Conc, lnConc
```

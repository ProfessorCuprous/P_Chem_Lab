```
#This code block carries out the linear fit to Model 3.

X = m
Y = frequency

from sklearn.linear_model import LinearRegression
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns; sns.set()

model = LinearRegression(fit_intercept=True)

model.fit(X[:,np.newaxis], Y)

Xfit = np.linspace(-10, 10, 10)
Yfit = model.predict(Xfit[:,np.newaxis])

plt.scatter(X, Y)        # These lines of code create a plot of the data (circles) alongside the linear fit (solid line).
plt.plot(Xfit, Yfit);    # This allows us to visually compare the agreement between the data and the model.
plt.xlabel("m")
plt.ylabel("Frequency")
plt.title("Frequency vs. m")
```

```
# This block of code provides the slope (shown in the table as 'm'), y-intercept (shown in the table as 'const').
# We can also see the uncertainties associated with these values.

from sklearn import linear_model
import statsmodels.api as sm
from pandas.testing import assert_frame_equal

# with statsmodels
X = sm.add_constant(X) # adding a constant

model = sm.OLS(Y, X).fit()
predictions = model.predict(X)

print_model = model.summary()
print(print_model)
```

```
# This code block plots the residuals.

X = m
Y = frequency

model = LinearRegression(fit_intercept=True)

model.fit(X[:,np.newaxis], Y)

Xfit = np.linspace(-10, 10, 10)
Yfit = model.predict(Xfit[:,np.newaxis])

Y_predicted=model.intercept_+model.coef_*X
residuals = Y-Y_predicted
plt.plot(X,residuals, 'o', color='darkblue')
plt.title("Residual Plot")
plt.xlabel("Independent Variable")
plt.ylabel("Residual")
```

In moving to Model 4, we will need values for m${^2}$ and m${^3}$. We can then carry out a multiple linear regression.

```
#This code block takes the array of "m" values and squares each of them, creating the array "m_squared".
#
#
m_squared=m*m                              #Square each of the m values
m_squared                                  #Print out the array to make sure we got it correct
```

```
#This code block takes the array of "m" values and cubes each of them, creating the array "m_cubed".
#
#
m_cubed = m*m*m                          #Cube each of hte m values
m_cubed                                  #Print out the array to make sure we got it correct
```

```
# This code block carries out the multiple linear regression.
# In the lines below, you will need to replace the word "HCl" with the name of your molecule if you are not using HCl.

HCl = {'frequency': frequency,
                'm': m,
                'm_squared': m_squared,
                'm_cubed': m_cubed
                }

dataPeaks = pd.DataFrame(HCl,columns=['frequency','m','m_squared','m_cubed'])

X = dataPeaks[['m','m_squared','m_cubed']]
Y = dataPeaks['frequency']

# with sklearn
regr = linear_model.LinearRegression()
regr.fit(X, Y)

print('Intercept: \n', regr.intercept_)
print('Coefficients: \n', regr.coef_)


# with statsmodels
X = sm.add_constant(X) # adding a constant

model = sm.OLS(Y, X).fit()
predictions = model.predict(X)

print_model = model.summary()    #These lines provide the output table below.
print(print_model)
```

```
from tabulate import tabulate
import math
# The code below uses the fit parameters and errors to calculate molecular parameters and uncertainties.
v=const_coef
v_un=const_std_err
alpha=-m_squared_coef
alpha_un=-m_squared_std_err
D=-m_cubed_coef/4
D_un=m_cubed_std_err/4
B=(m_coef+(2*alpha))/2
B_un=math.sqrt(((m_std_err*m_std_err)/4)+(alpha_un*alpha_un))
red_mass=34.9688*1.0078/(34.9688+1.0078)*1.66045E-27
I_e=(6.626E-34)/8/(math.pi*math.pi)/29980000000/B
I_e_un=I_e/B*B_un
I_e_lit=(6.626E-34)/8/(math.pi*math.pi)/29980000000/B_lit
r_e_m=math.sqrt(I_e/red_mass)
r_e_m_un=0.5*r_e_m/I_e*I_e_un
r_e_pm=r_e_m*1e12
r_e_pm_un=r_e_m_un*1e12
#
# The code below puts everything into a table.
d = [ ["\u03BD (cm\u207b\u00b9)", v, v_un, v_lit],
     ["\u03B1 (cm\u207b\u00b9)", alpha, alpha_un, alpha_lit],
     ["D (cm\u207b\u00b9)", D, D_un, D_lit],
     ["B (cm\u207b\u00b9)", B, B_un, B_lit],
     ["\u03BC (kg)", red_mass, 0, red_mass],
     ["I\u2091 (kg m\u00b2)", I_e, I_e_un, I_e_lit],
     ["r\u2091 (m)", r_e_m, r_e_m_un, r_e_m_lit],
     ["r\u2091 (pm)", r_e_pm, r_e_pm_un, r_e_pm_lit]]

print(tabulate(d, headers=["Parameter", "Experimental Value", "Uncertainty", "Literature Value"]))
```
